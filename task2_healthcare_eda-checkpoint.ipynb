```python
import pandas as pd

df = pd.read_csv("swynex_healthcare_hospital_cleaned.csv")
print("Rows & Columns:", df.shape)
df.head()
```

```python
cols = ["age", "billing_amount", "length_of_stay_days", "satisfaction_score"]
print(df[cols].describe().round(2))

readmit_rate = (df["readmission_flag"] == "Yes").mean() * 100
print(f"\nOverall Readmission Rate: {readmit_rate:.2f}%")
```

```python
cond_summary = (
    df.groupby("medical_condition")
    .agg(
        patients=("patient_id", "count"),
        avg_billing=("billing_amount", "mean"),
        avg_stay=("length_of_stay_days", "mean"),
        readmit_pct=("readmission_flag", lambda s: (s == "Yes").mean() * 100),
    )
    .round(2)
)
cond_summary
```

```python
df["month"] = pd.to_datetime(df["admission_date"]).dt.to_period("M").astype(str)
monthly_admissions = df.groupby("month")["patient_id"].count()

print(monthly_admissions)
```

```python
high_stay = df[df["length_of_stay_days"] > 20][[
    "patient_id",
    "medical_condition",
    "admission_date",
    "discharge_date",
    "length_of_stay_days",
    "billing_amount",
]]

print(f"Records with stay > 20 days: {len(high_stay)}")
high_stay
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(13, 8))

cond_summary["avg_billing"].sort_values().plot(
    kind="barh", ax=axes[0, 0], title="Avg Billing by Condition", color="#2b6cb0"
)
cond_summary["avg_stay"].sort_values().plot(
    kind="barh",
    ax=axes[0, 1],
    title="Avg Length of Stay (Days)",
    color="#2f855a",
)
monthly_admissions.plot(
    kind="line",
    marker="o",
    ax=axes[1, 0],
    title="Monthly Admissions Trend",
    color="#805ad5",
)
axes[1, 1].scatter(
    df["length_of_stay_days"],
    df["billing_amount"],
    alpha=0.5,
    color="#dd6b20",
)
axes[1, 1].set_title("Length of Stay vs. Billing Amount")
axes[1, 1].set_xlabel("Stay Days")
axes[1, 1].set_ylabel("Billing Amount")

plt.tight_layout()
plt.savefig("task2_healthcare_eda.png", dpi=150)
plt.show()
```

### 5 Key Insights from EDA

1. **High Average Billing in Pneumonia & Cardiac Care:** Pneumonia (`₹73,251.66`) and Cardiac Care (`₹71,785.28`) have higher average billing than other conditions like Arthritis (`₹53,999.56`) and Diabetes (`₹54,609.08`).

2. **Longer Average Stay in Hypertension:** Hypertension patients stay the longest on average (`6.05` days), whereas Diabetes patients have the shortest average stay (`4.41` days).

3. **Readmission Rate Variation:** Overall readmission rate is `22.59%`, with Asthma (`26.03%`) and Pneumonia (`24.14%`) showing relatively higher readmission shares than Hypertension (`19.86%`).

4. **Monthly Admissions Trend:** Admissions fluctuate across 2025, peaking in August (`2025-08` with `97` patients) and dipping lowest in March (`2025-03` with `56` patients).

5. **Stay Duration Outliers / Anomalies:** Most stays range between `1` and `15` days, while `5` records show extended stays (`22` to `157` days) that should be reviewed for clinical complexity or data accuracy.uracy.
